```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('dataset.csv')
BASE_PRICE = 10.0
```

```python
ALPHA = 2.0
data['price'] = BASE_PRICE
def compute_next_price(current_price, occupancy, capacity, alpha=ALPHA):
    if capacity == 0:
        return current_price
    occupancy_rate = occupancy / capacity
    next_price = current_price + alpha * occupancy_rate
    return round(next_price, 2)

data.sort_values(by=['lot_id', 'timestamp'], inplace=True)
def apply_pricing_model(group):
    prices = [BASE_PRICE]
    for i in range(1, len(group)):
        prev_price = prices[-1]
        row = group.iloc[i]
        next_price = compute_next_price(
            prev_price, row['occupancy'], row['capacity']
        )
        prices.append(next_price)
    group['price'] = prices
    return group

if 'lot_id' in data.columns:
    data = data.groupby('lot_id').apply(apply_pricing_model).reset_index(drop=True)
else:

    data = apply_pricing_model(data)

output_filename = 'baseline_model_output.csv'
data.to_csv(output_filename, index=False)
print(f"Output saved to {output_filename}")
```
